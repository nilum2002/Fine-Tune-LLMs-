```python
import os
import transformers
import torch
from google.colab import userdata
from datasets import load_dataset
from trl import SFTTrainer
from peft import LoraConfig
from transformers import AutoTokenizer, AutoModelForCausalLM # for generating some text based on decoder based transformer
from transformers import BitsAndBytesConfig, GemmaTokenizer
```

```python
from google.colab import userdata
import os


os.environ["HF_TOKEN"] = userdata.get("HF_TOKEN")
```

```python
import pandas as pd

splits = {'train': 'data/train-00000-of-00001.parquet', 'test': 'data/test-00000-of-00001.parquet'}
df = pd.read_parquet("hf://datasets/Thimira/sinhala-llm-dataset-llama-prompt-format/" + splits["train"])
```

```python
model_id = "google/gemma-2b-it"
bnb_config = BitsAndBytesConfig(
    load_in_4bit=True, # all 32 bit weights converts in to 4 bits
    bnb_4bit_quant_type="nf4", # 4-bit normal Float
    bnb_4bit_compute_dtype=torch.bfloat16
)
```

```python
tokenizer = AutoTokenizer.from_pretrained(model_id, token = os.environ["HF_TOKEN"])
model = AutoModelForCausalLM.from_pretrained(model_id, quantization_config=bnb_config,
                                             token = os.environ["HF_TOKEN"],
                                             device_map={"":0}
)
```

```python

lora_config = LoraConfig(
    r = 8,
    target_modules = ["q_proj", "o_proj", "k_proj", "v_proj", "gate_proj", "up_proj", "down_proj"],
    task_type = "CAUSAL_LM"
)
```

```python
def split_text(text):
    # Assuming the format is always '<s>[INST] prompt [/INST] response</s>'
    parts = text.split('[/INST]', 1)
    if len(parts) == 2:
        prompt_part = parts[0]
        response_part = parts[1]

        # Extracting the actual prompt after '<s>[INST]'
        prompt = prompt_part.replace('<s>[INST]', '').strip()

        # Extracting the actual response before '</s>'
        response = response_part.replace('</s>', '').strip()
        return prompt, response
    else:
        # Handle cases where the format might not match, or return original text
        return '', text # Or raise an error, or log a warning

df[['prompt', 'response']] = df['text'].apply(lambda x: pd.Series(split_text(x)))
df = df.drop(columns=['text'])
print(df.head())
```

```python
from datasets import DatasetDict, Dataset

dataset = Dataset.from_pandas(df)

splitted_dataset = dataset.train_test_split(test_size=0.1, seed=42)

data = DatasetDict({
    "train": splitted_dataset["train"],
    "test": splitted_dataset["test"]
})

print(data)
```

```python
def formatting_func_str(example):
    return f"<s>[INST] {example['prompt']} [/INST] {example['response']}</s>"

# Pre-process the dataset to create a 'text' column
data["train"] = data["train"].map(lambda example: {"text": formatting_func_str(example)}, remove_columns=["prompt", "response"])

trainer = SFTTrainer(
    model = model,
    train_dataset = data["train"],
    args = transformers.TrainingArguments(
        per_device_train_batch_size = 1,
        gradient_accumulation_steps = 4,
        warmup_steps = 2,
        max_steps = 100000,
        learning_rate = 2e-4,
        fp16 = False,
        logging_steps = 1,
        output_dir = "outputs",
        optim = "paged_adamw_8bit"
    ),
    peft_config = lora_config
    # The formatting_func is no longer needed here as the dataset is already pre-formatted to 'text'
    # SFTTrainer will look for the 'text' column by default.
)
```

```python
trainer.train()
```

```python
sinhala_question = "ෂ්‍රී ලංකාවේ ජනාදිපති කවුද ?" # What is the capital of Sri Lanka?
test_prompt = f"<s>[INST] {sinhala_question} [/INST]"
print(test_prompt)
```

```python
input_ids = tokenizer(test_prompt, return_tensors="pt").to(model.device)
outputs = model.generate(**input_ids, max_new_tokens=100)
response = tokenizer.decode(outputs[0], skip_special_tokens=True)

print("Original Prompt:")
print(test_prompt)
print("\nModel Response:")
print(response)
```

```python
test_examples = data["test"].select(range(5))

for example in test_examples:
    original_prompt = example['prompt']
    true_response = example['response']

    # Construct the input prompt for the model
    model_input_prompt = f"<s>[INST] {original_prompt} [/INST]"

    # Tokenize and generate response
    input_ids = tokenizer(model_input_prompt, return_tensors="pt").to(model.device)
    outputs = model.generate(**input_ids, max_new_tokens=100, do_sample=True, top_k=50, top_p=0.95)
    generated_response = tokenizer.decode(outputs[0][input_ids.input_ids.shape[1]:], skip_special_tokens=True)

    print(f"\n--- Example ---")
    print(f"Original Prompt: {original_prompt}")
    print(f"True Response: {true_response}")
    print(f"Model Generated Response: {generated_response}")
    print("----------------")
```

```python
import evaluate

# Load the BLEU metric
bleu = evaluate.load("bleu")

# Prepare lists for predictions and references
predictions = []
references = []

# Use the same test_examples from before or select more for a robust evaluation
# For demonstration, we'll use a small subset. You can increase this for a full evaluation.
test_examples_for_eval = data["test"].select(range(20))

for example in test_examples_for_eval:
    original_prompt = example['prompt']
    true_response = example['response']

    model_input_prompt = f"<s>[INST] {original_prompt} [/INST]"

    input_ids = tokenizer(model_input_prompt, return_tensors="pt").to(model.device)

    # Generate with some sampling to get more varied responses if do_sample is False
    outputs = model.generate(**input_ids, max_new_tokens=100, do_sample=True, top_k=50, top_p=0.95)
    generated_response = tokenizer.decode(outputs[0][input_ids.input_ids.shape[1]:], skip_special_tokens=True)

    predictions.append(generated_response)
    # BLEU expects references to be a list of lists (multiple possible references for each prediction)
    references.append([true_response])

# Compute the BLEU score
bleu_score = bleu.compute(predictions=predictions, references=references)

print(f"BLEU Score: {bleu_score}")
```
